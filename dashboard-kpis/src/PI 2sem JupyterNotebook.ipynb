```python
import pandas as pd
df = pd.read_csv("C:/Users/gabri/Downloads/KAGGLE/Simplificado - Final Dataset - State of Data 2024 - Kaggle - df_survey_2024.csv")

# Removendo espaços antes e depois de valores, espaços duplicados
def limpar_texto(texto):
    if pd.isna(texto):
        return texto
    texto = str(texto)
    texto = texto.strip()                # remove espaços no início e fim
    texto = ' '.join(texto.split())      # remove espaços duplicados internos
    return texto

df = df.map(limpar_texto)
```

```python
# Qual a faixa salarial dos profissionais da área?
ordem_salarios = [
    "Menos de 1.000",
    "de 1.001 a 2.000",
    "de 2.001 a 3.000",
    "de 3.001 a 4.000",
    "de 4.001 a 6.000",
    "de 6.001 a 8.000",
    "de 8.001 a 12.000",
    "de 12.001 a 16.000",
    "de 16.001 a 20.000",
    "de 20.001 a 25.000",
    "de 25.001 a 30.000",
    "de 30.001 a 40.000",
    "Acima de 40.001"
]
df['faixa_salarial_ordenada'] = pd.Categorical(
    df['2.h_faixa_salarial'],
    categories=ordem_salarios,
    ordered=True)

contagem_salario = df.groupby('faixa_salarial_ordenada', observed=False).size().reset_index(name='quantidade')

contagem_salario = contagem_salario.dropna()

contagem_salario.to_csv("C:/Users/gabri/Downloads/KAGGLE/data_sets/faixa_salarial.csv", index=False, encoding='utf-8-sig')
```

```python
# Quais tecnologias são mais utilizadas?
# Isolando IDs e a coluna com a informação
df_linguagens = df[['0.a_token', '4.d_linguagem_de_programacao_(dia_a_dia)']].copy()
df_linguagens = df_linguagens.dropna(subset=['4.d_linguagem_de_programacao_(dia_a_dia)'])

# Separar por vírgula e explodir
df_linguagens['linguagem'] = df_linguagens['4.d_linguagem_de_programacao_(dia_a_dia)'].str.split(',')
df_linguagens = df_linguagens.explode('linguagem')
df_linguagens['linguagem'] = df_linguagens['linguagem'].str.strip()

# Contando
contagem_linguagens = df_linguagens.groupby('linguagem').size().reset_index(name='quantidade')
contagem_linguagens = contagem_linguagens.sort_values('quantidade', ascending=False)

contagem_linguagens.to_csv("C:/Users/gabri/Downloads/KAGGLE/data_sets/contagem_linguagens.csv", index=False, encoding='utf-8-sig')
```

```python
# Como o nível de experiência impacta o salário?
df_salario_nivel = df[['2.g_nivel', '2.h_faixa_salarial']].dropna()

df_salario_nivel['faixa_ordenada'] = pd.Categorical(
    df_salario_nivel['2.h_faixa_salarial'],
    categories=ordem_salarios,
    ordered=True
)

crosstab_nivel_salario = pd.crosstab(
    df_salario_nivel['2.g_nivel'],
    df_salario_nivel['faixa_ordenada'],
    dropna=True
)

crosstab_nivel_salario.to_csv("C:/Users/gabri/Downloads/KAGGLE/data_sets/nivel_vs_salario_matriz.csv", encoding='utf-8-sig')
```

```python
# Quais são os cargos mais comuns na área de dados?
contagem_cargos = df['2.f_cargo_atual'].dropna().value_counts().reset_index()
contagem_cargos.columns = ['cargo', 'quantidade']

contagem_cargos.to_csv("C:/Users/gabri/Downloads/KAGGLE/data_sets/contagem_cargos.csv", index=False, encoding='utf-8-sig')
```

```python
# Os profissionais de dados que usam IA generativa ganham mais? 
# Criar categoria simplificada
def categoria_ia(texto):
    if pd.isna(texto):
        return 'Não respondeu'
    if 'gratuitas' in texto:
        return 'Usa IA gratuita'
    if 'pago do meu próprio bolso' in texto:
        return 'Usa IA paga (próprio bolso)'
    if 'empresa em que trabalho paga' in texto:
        return 'Usa IA paga (empresa)'
    if 'Copilot' in texto:
        return 'Usa Copilot'
    if 'Não utilizo nenhum' in texto:
        return 'Não usa IA'
    return 'Outros'

df['categoria_ia'] = df['4.m_usa_chatgpt_ou_copilot_no_trabalho?'].apply(categoria_ia)

# Remover valores vazios de salário
df_ia_salario = df[['categoria_ia', '2.h_faixa_salarial']].dropna(subset=['2.h_faixa_salarial'])

# Aplicar ordem salarial (já definida antes)
df_ia_salario['faixa_ordenada'] = pd.Categorical(
    df_ia_salario['2.h_faixa_salarial'],
    categories=ordem_salarios,
    ordered=True
)

# Tabela cruzada
crosstab_ia_salario = pd.crosstab(
    df_ia_salario['categoria_ia'],
    df_ia_salario['faixa_ordenada'],
    dropna=True
)

crosstab_ia_salario.to_csv("C:/Users/gabri/Downloads/KAGGLE/data_sets/ia_vs_salario.csv", encoding='utf-8-sig')
```
